## Data Preprocessing and Cleaning

```python
# Remove columns that are not required for the planned business analysis.
df.drop(columns=['Zone', 'Sharing_Capacity', 'Amenities', 'Check_In_Date'], inplace=True)
df.columns
```

```python
# Replace missing electricity charges with 0 because no recorded charge is treated as zero.
df['Electricity_Charges'] = df['Electricity_Charges'].fillna(0)
df['Water_Charges'] = df['Water_Charges'].fillna(0)
df['Food_Type'] = df['Food_Type'].fillna(df['Food_Type'].mode()[0])
df['Nearest_Metro'] = df['Nearest_Metro'].fillna(df['Nearest_Metro'].mode()[0])
df['Review_Count'] = df['Review_Count'].fillna(df['Review_Count'].median())
df['Distance_to_Metro_km'] = df['Distance_to_Metro_km'].fillna(df['Distance_to_Metro_km'].median())
```

```python
# Round Google ratings to one decimal place for a consistent presentation format.
invalid_rating = (df['Google_Rating'] < 0) | (df['Google_Rating'] > 5)
print("Invalid rating rows:", invalid_rating.sum())

df.loc[invalid_rating, 'Google_Rating'] = np.nan
df['Google_Rating'] = df['Google_Rating'].fillna(df['Google_Rating'].median()).round(1)

print("New range:", df['Google_Rating'].min(), "-", df['Google_Rating'].max())


```

```python
df['Nearest_IT_Park'] = df['Nearest_IT_Park'].astype(str).str.strip()
df['Nearest_IT_Park'] = df['Nearest_IT_Park'].replace('', np.nan)
df['Nearest_IT_Park'] = df['Nearest_IT_Park'].fillna(df['Nearest_IT_Park'].mode()[0])

print("Blank Nearest_IT_Park rows remaining:", (df['Nearest_IT_Park'].str.strip()=='').sum())
```

```python
# Recheck missing values after the initial missing-value treatment.
df.isnull().sum()
```

```python
# Convert invalid/missing Security Deposit values
df['Security_Deposit'] = df['Security_Deposit'].replace({
    'Nil': 0,
    '2 Months': 0
})

# Standardize invalid Monthly Rent values
df['Monthly_Rent'] = df['Monthly_Rent'].replace({
    'Rs 999999': 999999,
    '999999/month': 999999,
    'â\x82¹999999': 999999,
    '9,99,999': 999999,
    -9999: 999999
})
```

```python
# Create a mapping to standardize different representations of Yes/No values.
mapping = {
    '0': 'No',
    'NO': 'No',
    'yes': 'Yes',
    'Y': 'Yes',
    'no': 'No',
    'False': 'No',
    'True': 'Yes',
    'YES': 'Yes',
    'yes': 'Yes',
    'N': 'No',
    '1': 'Yes'
}
amenity_columns = [
    'WiFi', 'Laundry', 'Parking', 'Power_Backup', 'Housekeeping',
    'Gym', 'Lift', 'CCTV', 'RO_Water', 'AC', 'TV', 'Study_Table',
    'Geyser', 'Washing_Machine'
]
df[amenity_columns] = df[amenity_columns].replace(mapping)
```

```python
# Convert security deposit to integer for numerical calculations.
df['Security_Deposit'] = df['Security_Deposit'].astype('int64')
df['Electricity_Charges'] = df['Electricity_Charges'].astype('int64')
df['Water_Charges'] = df['Water_Charges'].astype('int64')
df['Monthly_Rent'] = df['Monthly_Rent'].astype('int64')
df['Review_Count'] = df['Review_Count'].astype('int64')
df.dtypes
```

```python
# Fill missing locality names with the most frequently occurring locality.
df['Area'] = df['Area'].fillna(df['Area'].mode()[0])
```

```python
# Import Python's regular-expression module for text processing.
df['Area'] = df['Area'].str.strip()
import re
from collections import defaultdict

def _key(a):
    return re.sub(r'\s+', '', a).lower()

_groups = defaultdict(list)
for a in df['Area'].unique():
    _groups[_key(a)].append(a)

_area_map = {}
for k, variants in _groups.items():
    single_space = [v for v in variants if '  ' not in v]
    spaced = [v for v in single_space if ' ' in v]
    canonical = spaced[0] if spaced else (single_space[0] if single_space else variants[0])
    for v in variants:
        _area_map[v] = canonical

print('Area categories before fix:', df['Area'].nunique())
df['Area'] = df['Area'].map(_area_map)
print('Area categories after fix:', df['Area'].nunique())

```

### 3) Outlier detection — every key numeric column

```python
fig, axes = plt.subplots(3, 4, figsize=(20, 12))
axes = axes.flatten()

for i, col in enumerate(numeric_cols):
    data = rent_eda[col] if col == 'Monthly_Rent' else df[col]
    sns.boxplot(y=data, ax=axes[i])
    axes[i].set_title(col)

plt.tight_layout()
plt.show()
```

```python
outlier_summary = []

for col in numeric_cols:
    data = rent_eda[col] if col == 'Monthly_Rent' else df[col]
    Q1, Q3 = data.quantile(0.25), data.quantile(0.75)
    IQR = Q3 - Q1
    lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
    n_outliers = ((data < lower) | (data > upper)).sum()
    outlier_summary.append({
        'Column': col,
        'Lower Bound': round(lower, 1),
        'Upper Bound': round(upper, 1),
        'Outlier Count': n_outliers,
        'Outlier %': round(100 * n_outliers / len(data), 2)
    })

outlier_df = pd.DataFrame(outlier_summary).sort_values('Outlier %', ascending=False)
outlier_df
```

**Observation:** 
- Electricity_Charges has the most outliers (~3.9%, mostly high-end values above 
₹1875) followed by Monthly_Rent (~2%) and Security_Deposit (~1.2%) — these are real, plausible 
values, not data errors (already handled the true sentinel/invalid values during cleaning). 
- Water_Charges, Maintenance_Fee, Review_Count, Distance_to_Metro_km, Distance_to_IT_Park_km, 
Safety_Score, Commute_Score, and Value_for_Money show 0 IQR outliers — very well-behaved, bounded

## Modeling: Clustering + Classification

Segment PGs into meaningful groups (KMeans), then train a classifier that can assign any new PG 
listing to the right group instantly — this is what powers the Streamlit app's prediction.

```python
from sklearn.preprocessing import RobustScaler, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             classification_report, confusion_matrix, ConfusionMatrixDisplay)
import joblib
```

### Clustering ###

### 1. K-Means Clustering ###


1.1 Feature selection

```python
# ================= 1. K-MEANS CLUSTERING =================

# 1.1 Feature selection
base_features = ['Monthly_Rent', 'Security_Deposit', 'Google_Rating', 'Review_Count',
                 'Distance_to_Metro_km', 'Distance_to_IT_Park_km',
                 'Safety_Score', 'Commute_Score', 'Value_for_Money']
df[base_features].describe().round(2)
```

1.2 Feature engineering

```python
amenity_cols = ['WiFi', 'Laundry', 'Parking', 'Power_Backup', 'Housekeeping', 'Gym', 'Lift',
                'CCTV', 'RO_Water', 'AC', 'TV', 'Study_Table', 'Geyser', 'Washing_Machine']

df['Amenity_Count'] = df[amenity_cols].eq('Yes').sum(axis=1)
df['Total_Monthly_Cost'] = (df['Monthly_Rent'] + df['Electricity_Charges']
                            + df['Water_Charges'] + df['Maintenance_Fee'])

df[['Monthly_Rent', 'Total_Monthly_Cost', 'Amenity_Count']].head()
```

1.3 Remove invalid rows

```python
print("Sentinel rent rows:", (df['Monthly_Rent'] == 999999).sum())
model_df = df[df['Monthly_Rent'] < 50000].copy().reset_index(drop=True)
print("Before:", df.shape, "| After:", model_df.shape)
```

1.4 Preprocessing

```python
skewed_cols   = ['Monthly_Rent', 'Security_Deposit', 'Total_Monthly_Cost']
standard_cols = ['Google_Rating', 'Review_Count', 'Distance_to_Metro_km', 'Distance_to_IT_Park_km',
                 'Safety_Score', 'Commute_Score', 'Value_for_Money', 'Amenity_Count']
features = skewed_cols + standard_cols
X = model_df[features].copy()
X.isnull().sum()
```

1.5 — Scaling

```python
preprocessor = ColumnTransformer(transformers=[
    ('robust',   RobustScaler(),   skewed_cols),
    ('standard', StandardScaler(), standard_cols)
])
X_scaled = preprocessor.fit_transform(X)
print("Scaled feature matrix:", X_scaled.shape)
```

1.6 — Elbow method


```python
K_range = range(2, 11)
inertias = []
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled)
    inertias.append(km.inertia_)

plt.figure(figsize=(9, 5))
plt.plot(list(K_range), inertias, marker='o', linewidth=2)
plt.title("Elbow Method — Inertia vs Number of Clusters")
plt.xlabel("Number of clusters (k)")
plt.ylabel("Inertia (WCSS)")
plt.xticks(list(K_range))
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Observation:**
- No sharp elbow is visible; inertia decreases smoothly across all k. This suggests the data lacks a single obvious natural cluster count.
- After K=4, increasing the number of clusters gives diminishing returns.
- we can take values after 4 also but the changes are small and the most change is coming at 4 so thats why from elbow we will choose 4.

1.7 — Silhouette score

```python
sil_scores = []
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    sil_scores.append(silhouette_score(X_scaled, labels))

plt.figure(figsize=(9, 5))
plt.plot(list(K_range), sil_scores, marker='o', color='green', linewidth=2)
plt.title("Silhouette Score vs Number of Clusters")
plt.xlabel("Number of clusters (k)")
plt.ylabel("Silhouette Score")
plt.xticks(list(K_range))
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

best_k = list(K_range)[int(np.argmax(sil_scores))]
print("Best k by silhouette:", best_k, "| score:", round(max(sil_scores), 4))
```

**Observation:**
- Silhouette peaks at k=2 (~0.08) then stays low and flat, meaning no strongly separated clusters exist. This matches the near-zero correlations found during EDA.

1.8 — Select K

Why: Statistically k=2 scores highest, but that just splits into "cheap vs expensive" — not very useful. k=4 scores only marginally lower but gives four distinct, nameable groups a student can actually use. When the stats are this flat, business usefulness wins.

```python
FINAL_K = 4
print("Selected number of clusters:", FINAL_K)
print("Silhouette at k=2:", round(sil_scores[0], 4))
print("Silhouette at k=4:", round(sil_scores[2], 4))
```

1.9 — Final K-Means

```python
kmeans = KMeans(n_clusters=FINAL_K, random_state=42, n_init=10)
model_df['Cluster'] = kmeans.fit_predict(X_scaled)
print("Silhouette:", round(silhouette_score(X_scaled, model_df['Cluster']), 4))
print("Inertia:", round(kmeans.inertia_, 2))
model_df[['PG_Name', 'Area', 'Monthly_Rent', 'Safety_Score', 'Cluster']].head(10)
```

**Observation:**
- The model converges with a silhouette score around 0.07, confirming weak but usable separation. Cluster sizes are reasonably balanced, not dominated by one group.

- Rent, deposit, and total cost differ the most across clusters, while amenities barely change. Safety and commute scores are what split the remaining lower-cost clusters apart.

1.12 — PCA visualization

```python
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
print("Variance explained:", np.round(pca.explained_variance_ratio_, 4))

plt.figure(figsize=(10, 7))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=model_df['Cluster'], cmap='tab10', alpha=0.6, s=18)
centroids_pca = pca.transform(kmeans.cluster_centers_)
plt.scatter(centroids_pca[:, 0], centroids_pca[:, 1], c='black', marker='X', s=250, label='Centroids')
plt.title("PG Clusters — PCA 2D Projection")
plt.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)")
plt.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)")
plt.legend(*scatter.legend_elements(), title="Cluster")
plt.tight_layout()
plt.show()
```

**Observation:**
- The two components explain only ~23% of variance, so the plot understates true separation. Overlap in 2D does not mean the clusters are invalid in the full feature space.

1.13 — Observation (overall)

The clustering is valid but weakly separated (silhouette ~0.07), consistent with the near-zero correlations found in EDA. The four segments are still business-meaningful, mainly separated by rent/cost, Safety_Score, and Commute_Score.

### Classification ###

2.1 — Define target

```python
target = 'Cluster'
print("Target:", target)
print("Classes:", model_df[target].nunique())
print(model_df[target].value_counts().sort_index())
```

2.2 — X / y

```python
X_clf = model_df[features].copy()
y_clf = model_df[target].copy()
print("X:", X_clf.shape, "| y:", y_clf.shape)
X_clf.head()
```

2.3 — Train-test split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_clf, y_clf, test_size=0.2, random_state=42, stratify=y_clf
)
print("Train:", X_train.shape, "| Test:", X_test.shape)
```

2.4 — Preprocessing

```python
clf_preprocessor = ColumnTransformer(transformers=[
    ('robust',   RobustScaler(),   skewed_cols),
    ('standard', StandardScaler(), standard_cols)
])
X_train_scaled = clf_preprocessor.fit_transform(X_train)
X_test_scaled  = clf_preprocessor.transform(X_test)

results = []
trained_models = {}

def evaluate(name, model, y_pred):
    results.append({
        'Model'    : name,
        'Accuracy' : accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred, average='macro'),
        'Recall'   : recall_score(y_test, y_pred, average='macro'),
        'F1-Score' : f1_score(y_test, y_pred, average='macro')
    })
    trained_models[name] = model
    print(f"{name} — Accuracy: {accuracy_score(y_test, y_pred):.4f}")
    print(classification_report(y_test, y_pred))
```

2.5 — KNN (baseline)

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)
y_pred_knn = knn.predict(X_test_scaled)
evaluate("KNN", knn, y_pred_knn)
```

2.6 — Logistic Regression

```python
logreg = LogisticRegression(max_iter=1000, random_state=42)
logreg.fit(X_train_scaled, y_train)
y_pred_logreg = logreg.predict(X_test_scaled)
evaluate("Logistic Regression", logreg, y_pred_logreg)
```

2.7 — Random Forest

```python
rf = RandomForestClassifier(n_estimators=200, random_state=42)
rf.fit(X_train_scaled, y_train)
y_pred_rf = rf.predict(X_test_scaled)
evaluate("Random Forest", rf, y_pred_rf)

importances = pd.Series(rf.feature_importances_, index=features).sort_values()
plt.figure(figsize=(9, 6))
importances.plot(kind='barh')
plt.title("Random Forest — Feature Importance")
plt.tight_layout()
plt.show()
```

2.8 — Compare metrics

```python
results_df = pd.DataFrame(results).set_index('Model').round(4)
results_df = results_df.sort_values('Accuracy', ascending=False)
print(results_df)

best_model_name = results_df.index[0]
print(f"\nBest model: {best_model_name} ({results_df.loc[best_model_name, 'Accuracy']:.2%})")

ax = results_df.plot(kind='bar', figsize=(12, 6), rot=20, width=0.8)
ax.set_title("Classification Model Comparison")
ax.set_ylim(0, 1.05)
plt.tight_layout()
plt.show()
```

**Observation:**
- Logistic Regression clearly outperforms Random Forest and KNN. This gap points to the cluster boundaries being close to linear, which Logistic Regression can capture directly.

2.9 — Confusion matrix

```python
predictions = {
    'KNN'                : y_pred_knn,
    'Logistic Regression': y_pred_logreg,
    'Random Forest'      : y_pred_rf
}

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for i, (name, y_pred) in enumerate(predictions.items()):
    cm = confusion_matrix(y_test, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False, ax=axes[i])
    axes[i].set_title(f"{name}\nAccuracy: {accuracy_score(y_test, y_pred):.3f}")
    axes[i].set_xlabel("Predicted"); axes[i].set_ylabel("Actual")
plt.tight_layout()
plt.show()

best_pred = predictions[best_model_name]
print(f"Best model: {best_model_name}\n")
print(classification_report(y_test, best_pred))
```

**Observation:**
- Errors are concentrated between adjacent, similarly-priced clusters rather than spread randomly. The premium cluster is almost never confused with any budget cluster.

2.10 — Overall Observation

Logistic Regression performs best because K-Means creates linear cluster boundaries that it recovers almost exactly, while KNN and Random Forest lag slightly since they approximate boundaries less directly. Feature importance confirms rent, Safety_Score, and Commute_Score drive the segmentation, matching Part 1's profile.
